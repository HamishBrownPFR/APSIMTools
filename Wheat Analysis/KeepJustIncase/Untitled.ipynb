```python
import datetime
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data = pd.read_clipboard()
```

```python
data.dropna(axis=1,how='all',inplace=True)
```

```python
num_cols = data.select_dtypes(include=np.number).columns
str_cols = data.select_dtypes(exclude=np.number).columns
```

```python
str_cols
```

```python
AgrigatedData = data.groupby(['SimulationName','Clock.Today']).agg({**{col: 'mean' for col in num_cols},
            **{col: 'first' for col in str_cols}})
```
