```python
import pandas as pd
import matplotlib.pyplot as plt

# Set up dataframe to parse test names and durations into
results = pd.DataFrame(columns=["FailedTooSlow","RecentlyPassed"])

# Read in results from log of test that terminated becasue took too long
file1 = open('FailedTooSlow.txt', 'r')
Lines = file1.readlines()
for l in Lines:
    try:
        test = l.split(" has finished. Elapsed time was ")[0]
        durat = l.split(" has finished. Elapsed time was ")[1].split(" ")[0]
        results.loc[test,"FailedTooSlow"] = durat
    except:
        do = 'nothing'

#Sort index        
results.sort_index(inplace=True)

# Read in results from log of test that recently passed
file2 = open('RecentlyPassed.txt', 'r')
Lines = file2.readlines()
for l in Lines:
    try:
        test = l.split(" has finished. Elapsed time was ")[0]
        durat = l.split(" has finished. Elapsed time was ")[1].split(" ")[0]
        results.loc[test,"RecentlyPassed"] = durat
    except:
        do = 'nothing'
        
#Show dataframe with simulations that recoreded a pass and duration in the recent log but not in the failed log
results.loc[results.FailedTooSlow.isna(),:]
```
